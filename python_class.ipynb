```python
n=int(input("Enter total monkeys(n):"))
k=int(input("Enter bananas a monkey can eat(k):"))
j=int(input("Enter peanuts a monkey can eat(j):"))
m=int(input("Enter total bananas(m):"))
p=int(input("Enter total peanuts(p):"))
if(k==0 or j==0):
  print("invalid input")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%j!=0):
    n=n-1
  print("No of monkeys in tree:",n)
```

```python
n=int(input("enter no:of candies available in the jar"))
k=int(input("enter number of minimum candies that must be inside JAR ever."))
C=1
R=n
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>n):
    print("invalid")
  else:
    R=R-M
    if(R<=k):
      R=n
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```
